```python
import json, random
from pathlib import Path

rows = [
    json.loads(line)
    for line in Path(DATASET_PATH).read_text(encoding="utf-8").splitlines()
    if line.strip()
]

random.seed(42)
random.shuffle(rows)

tiny = []
for i, row in enumerate(rows[:20]):
    if i < 10:
        row["split"] = "train"
    elif i < 16:
        row["split"] = "val"
    else:
        row["split"] = "test"
    tiny.append(row)

TINY_DATASET_PATH = "/kaggle/working/gepa_dataset_tiny.jsonl"

with open(TINY_DATASET_PATH, "w", encoding="utf-8") as f:
    for row in tiny:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print("Wrote:", TINY_DATASET_PATH)
```
